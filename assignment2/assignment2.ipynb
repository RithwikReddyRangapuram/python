```python
#Q3
import json
from datetime import datetime

payload = {
    "created_at": datetime(2026, 9, 16, 10, 30),
    "tags": {"python", "json", "files"}
}
try:
    print(json.dumps(payload))
except TypeError as e:
    print("Normal dumps:", e)


def json_default(obj):
    if isinstance(obj, datetime):
        return obj.isoformat()

    if isinstance(obj, set):
        return sorted(obj)

    raise TypeError(
        "Object of type {} is not JSON serializable".format(
            type(obj).__name__
        )
    )


result = json.dumps(
    payload,
    default=json_default,
    indent=2
)

print(result)
```

```python
#Q4
import json
from dataclasses import dataclass, asdict


@dataclass
class User:
    name: str
    age: int
user1 = User("Rithwik", 22)

json_data = json.dumps(asdict(user1))

print("JSON:", json_data)
data = json.loads(json_data)
user2 = User(**data)

print("User:", user2)

print("Round-trip worked:", user1 == user2)
```

```python
#Q7
import json
import os
import tempfile
from pathlib import Path
from typing import Any


def load_json(path: Path) -> Any:

    if not path.exists():
        return {}

    try:
        with path.open("r", encoding="utf-8") as file:
            return json.load(file)

    except json.JSONDecodeError as e:
        raise ValueError(
            "Invalid JSON in {}".format(path)
        ) from e


def save_json(path: Path, data: Any) -> None:

    path.parent.mkdir(
        parents=True,
        exist_ok=True
    )

    fd, temp_name = tempfile.mkstemp(
        prefix=path.name + ".",
        suffix=".tmp",
        dir=str(path.parent)
    )

    temp_path = Path(temp_name)

    try:
        with os.fdopen(
            fd,
            "w",
            encoding="utf-8"
        ) as file:

            json.dump(
                data,
                file,
                indent=2,
                ensure_ascii=False
            )

            file.flush()
            os.fsync(file.fileno())

        # Replace original with complete temporary file
        temp_path.replace(path)

    finally:
        if temp_path.exists():
            temp_path.unlink()


# --------------------
# Demonstration
# --------------------

folder = Path("q7_demo")

missing = folder / "missing.json"
corrupt = folder / "corrupt.json"
good = folder / "good.json"

folder.mkdir(exist_ok=True)

# 1. Missing file
if missing.exists():
    missing.unlink()

print("Missing:", load_json(missing))


# 2. Corrupt file
corrupt.write_text(
    "{oops",
    encoding="utf-8"
)

try:
    load_json(corrupt)
except ValueError as e:
    print("Corrupt:", e)


# 3. Successful round-trip
data = {
    "name": "Priya",
    "age": 22
}

save_json(good, data)

print(
    "Round-trip:",
    load_json(good) == data
)
```

```python
#Q8
from pathlib import Path

root = Path("_python")

if not root.exists():
    print("Folder does not exist:", root.resolve())
else:
    count = 0
    total_size = 0

    for file in root.rglob("*.py"):
        size = file.stat().st_size

        print(
            "{} - {} bytes".format(
                file,
                size
            )
        )

        count += 1
        total_size += size

    print("Total count:", count)
    print("Total size:", total_size, "bytes")
```
