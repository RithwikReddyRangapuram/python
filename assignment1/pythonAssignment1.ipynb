```python
import itertools
def every_nth(iterable, n):
    for index, item in enumerate(iterable):
        if index % n == 0:
            yield item

numbers = range(10**12)
generator = every_nth(numbers, 100)
first_three = itertools.islice(generator, 3)
print(list(first_three))
```

```python
from dataclasses import dataclass

@dataclass(frozen=True)
class Money:

    amount: float
    currency: str
    def __add__(self, other):

        if self.currency != other.currency:
            raise ValueError(
                "Currencies must match"
            )

        return Money(
            self.amount + other.amount,
            self.currency
        )


    def __str__(self):

        return f"{self.currency}{self.amount:,.2f}"

m1 = Money(1500, "₹")
m2 = Money(500, "₹")
total = m1 + m2
print(total)
try:
    m1.amount = 2000
except Exception as e:
    print(type(e).__name__)
```

```python
#Q18
from contextlib import contextmanager
import time


@contextmanager
def timer():
    start = time.perf_counter()

    try:
        yield
    finally:
        end = time.perf_counter()
        print(f"Time taken: {end - start:.6f} seconds")


with timer():
    sum(range(10_000_000))
```

```python
#Q24
def once(fn):
    called = False
    result = None

    def wrapper():
        nonlocal called, result

        if not called:
            result = fn()
            called = True

        return result

    return wrapper


counter = 0


def increment():
    global counter
    counter += 1
    print("Function called")
    return counter


run_once = once(increment)

print(run_once())
print(run_once())
print(run_once())

print("Counter:", counter)
```

```python
#q25
import heapq

users = [
    {"name": "Zoya", "age": 25, "salary": 60},
    {"name": "Aman", "age": 30, "salary": 90},
    {"name": "Ravi", "age": 25, "salary": 75},
]

# Age descending, name ascending
sorted_users = sorted(
    users,
    key=lambda user: (-user["age"], user["name"])
)

print(sorted_users)

# Two highest-paid users without sorting the entire list
highest_paid = heapq.nlargest(
    2,
    users,
    key=lambda user: user["salary"]
)

print(highest_paid)
```

```python
### Mini Project
from collections import Counter, defaultdict
from contextlib import contextmanager


LOGS = [
    "2026-09-09 10:01:22 INFO /home 200 0.031",
    "2026-09-09 10:01:25 ERROR /api/users 500 1.202",
    "2026-09-09 10:02:01 INFO /login 200 0.089",
    "2026-09-09 10:02:14 WARN /admin 403 0.011",
    "2026-09-09 10:02:20 ERROR /api/users 500 0.977",
    "2026-09-09 10:03:05 INFO /home 200 0.025",
    "bad line that should not crash the parser",
    "2026-09-09 10:03:40 INFO /api/orders 201 0.310",
]


class ParseError(Exception):
    pass


def parse_log(line):
    parts = line.split()

    if len(parts) != 6:
        raise ParseError("Malformed log line")

    date, clock, level, path, status, seconds = parts

    try:
        status = int(status)
        seconds = float(seconds)
    except ValueError:
        raise ParseError("Invalid status or response time")

    return {
        "time": f"{date} {clock}",
        "level": level,
        "path": path,
        "status": status,
        "seconds": seconds,
    }


@contextmanager
def banner(title):
    print(f"=== {title} ===")

    try:
        yield
    finally:
        print("=== end ===")


def main():
    parsed_logs = []
    bad_lines = 0

    for line in LOGS:
        try:
            parsed_logs.append(parse_log(line))
        except ParseError as error:
            print("Skipped:", error)
            bad_lines += 1

    # 1. Count requests by level
    level_counts = Counter(log["level"] for log in parsed_logs)

    print("\nRequests per level:")
    print(dict(level_counts))

    # 2. Group response times by path
    response_times = defaultdict(list)

    for log in parsed_logs:
        response_times[log["path"]].append(log["seconds"])

    # Dict comprehension for average response time
    averages = {
        path: sum(times) / len(times)
        for path, times in response_times.items()
    }

    # Sort slowest first
    sorted_averages = sorted(
        averages.items(),
        key=lambda item: item[1],
        reverse=True
    )

    print("\nAverage response time per path:")

    for path, average in sorted_averages:
        print(f"{path}: {average:.3f}s")

    # 3. Paths with 5xx responses
    error_paths = {
        log["path"]
        for log in parsed_logs
        if 500 <= log["status"] <= 599
    }

    print("\nPaths with 5xx status:")
    print(error_paths)

    # 4. Bad lines
    print("\nBad lines:", bad_lines)


if __name__ == "__main__":
    with banner("Log Analyzer"):
        main()
```
