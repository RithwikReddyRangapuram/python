```python
#Q2) asyncio.gather() is used to run multiple coroutines concurrently. It takes in multiple coroutine objects and returns a single coroutine that aggregates the results of all the input coroutines. When you call asyncio.gather(), it schedules all the input coroutines to run concurrently and waits for them to complete. Once all the coroutines have completed, it returns a list of their results in the order they were passed to gather().
import asyncio
import time
async def fetch(name, seconds):
    await asyncio.sleep(seconds)
    return f"{name} done"
async def sequential():
    result1 = await fetch("a", 1)
    result2 = await fetch("b", 1)

    return result1, result2
async def concurrent():
    return await asyncio.gather(
        fetch("a", 1),
        fetch("b", 1)
    )


async def main():
    start = time.perf_counter()
    print(await sequential())
    sequential_time = time.perf_counter() - start

    start = time.perf_counter()
    print(await concurrent())
    concurrent_time = time.perf_counter() - start

    print(f"Sequential: {sequential_time:.2f} seconds")
    print(f"Concurrent: {concurrent_time:.2f} seconds")


await main()
```

```python
#Q3
import asyncio
async def success(name):
    await asyncio.sleep(0.1)
    return f"{name} success"
async def failure():
    await asyncio.sleep(0.1)
    raise ValueError("boom")
async def main():
    print("Default gather:")

    try:
        result = await asyncio.gather(
            success("a"),
            failure(),
            success("c")
        )
        print(result)
    except ValueError as e:
        print("Caught:", e)

    print("\nreturn_exceptions=True:")

    result = await asyncio.gather(
        success("a"),
        failure(),
        success("c"),
        return_exceptions=True
    )

    print(result)
    print("Failed result type:", type(result[1]).__name__)


await main()
```

```python
#Q4
import asyncio
import time


async def job(number, semaphore):
    async with semaphore:
        print(f"Job {number} started")
        await asyncio.sleep(1)
        print(f"Job {number} finished")
        return number


async def main():
    semaphore = asyncio.Semaphore(2)

    start = time.perf_counter()

    await asyncio.gather(
        *(job(i, semaphore) for i in range(1, 7))
    )

    elapsed = time.perf_counter() - start

    print(f"Total time: {elapsed:.2f} seconds")


await main()
```

```python
#Q11)
import re


def normalize_phone(raw):
    # Remove spaces, hyphens, parentheses, etc.
    cleaned = re.sub(r"[\s\-()]", "", raw)

    # Remove Indian country code
    if cleaned.startswith("+91"):
        cleaned = cleaned[3:]
    elif cleaned.startswith("91") and len(cleaned) == 12:
        cleaned = cleaned[2:]

    # Remove leading zero
    if cleaned.startswith("0"):
        cleaned = cleaned[1:]

    # Validate exactly 10 digits starting from 6-9
    if re.fullmatch(r"[6-9]\d{9}", cleaned):
        return cleaned

    return None


numbers = [
    "98765 43210",
    "+91-9876543210",
    "09876543210",
    "12345",
    "98765-123"
]

for number in numbers:
    print(number, "->", normalize_phone(number))

#search can find a match anywhere in the string, while fullmatch requires the entire string to match the pattern. In this case, we want to ensure that the entire cleaned number is valid, so fullmatch is appropriate.
```

```python
###--MINI-PROJECT--
import asyncio
import logging
import random
import re
import time
from collections import Counter
from datetime import datetime, timezone


SITES = [
    "https://example.com",
    "https://api.example.com/health",
    "not a url",
    "https://shop.example.com",
    "ftp://old.example.com",
    "https://blog.example.com",
]


logging.basicConfig(
    level=logging.INFO,
    format="%(asctime)s - %(levelname)s - %(message)s"
)


def is_valid_url(url):
    pattern = r"https://[A-Za-z0-9.-]+\.[A-Za-z]{2,}"

    return re.fullmatch(pattern, url) is not None


async def check_site(url, semaphore):
    async with semaphore:
        seconds = random.uniform(0.1, 1.0)

        await asyncio.sleep(seconds)

        status = random.choice([
            200,
            200,
            200,
            500,
            503
        ])

        return {
            "url": url,
            "status": status,
            "seconds": seconds,
            "checked_at": datetime.now(timezone.utc).isoformat()
        }


async def main():
    random.seed(1)

    valid_sites = []

    for site in SITES:
        if is_valid_url(site):
            valid_sites.append(site)
        else:
            logging.warning("Invalid URL skipped: %s", site)

    semaphore = asyncio.Semaphore(2)

    start = time.perf_counter()

    results = await asyncio.gather(
        *(
            check_site(site, semaphore)
            for site in valid_sites
        )
    )

    wall_time = time.perf_counter() - start

    # Log each result
    for result in results:
        logging.info(
            "%s - status=%s - %.3fs",
            result["url"],
            result["status"],
            result["seconds"]
        )

    # Count statuses
    status_counts = Counter(
        result["status"]
        for result in results
    )

    print("\nStatus counts:")
    print(dict(status_counts))

    # Slowest site
    slowest = max(
        results,
        key=lambda result: result["seconds"]
    )

    print("\nSlowest site:")
    print(slowest["url"])
    print(f"Response time: {slowest['seconds']:.3f}s")

    # Total individual response times
    total_individual_time = sum(
        result["seconds"]
        for result in results
    )

    print(f"\nWall-clock time: {wall_time:.3f}s")
    print(f"Sum of individual times: {total_individual_time:.3f}s")

    print(
        "\nThe difference shows that the sites were checked "
        "concurrently rather than one after another."
    )


if __name__ == "__main__":
    await(main())
```
